```python
import tomllib

import polars as pl
```

```python
with open("config.toml", "rb") as fp:
    config = tomllib.load(fp)
```

```python
data = pl.read_parquet(f"{config['dir']}/budapest/homes/v2/home_work_h3_9.parquet")
```

```python
cleaned = (
    data.drop(["device_id", "count_home", "count_work"])
    .drop_nulls()
    .unique()
    .filter(pl.col.h3_9_home != pl.col.h3_9_work)
)
cleaned.columns = ["source", "target"]
```

```python
cleaned.write_csv("../data/od/picwell_hw_v2.csv")
```
